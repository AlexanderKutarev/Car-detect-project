```python
import cv2

# Загружаем исходное изображение
img = cv2.imread('C:\\Users\\sasha\\Desktop\\Car detect\\эксперименты!!!\\gleb on street.jpg')

# Перебираем все обнаруженные объекты
for box in results[0].boxes:
    
    if box.cls != 10:
        continue
    print(1)
    # Координаты bounding box (формат xyxy)
    x1, y1, x2, y2 = map(int, box.xyxy[0].tolist())
    
    # Класс объекта и его уверенность
    class_id = int(box.cls[0].item())
    confidence = box.conf[0].item()
    label = f"{model.names[class_id]} {confidence:.2f}"

    # Рисуем прямоугольник
    cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)  # (B, G, R), толщина
    
    # Добавляем текст (класс + уверенность)
    cv2.putText(img, label, (x1, y1 - 10), 
               cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

```
